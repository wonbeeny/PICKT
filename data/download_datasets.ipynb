```python
import pandas as pd

from pathlib import Path
from huggingface_hub import snapshot_download

repo_id = "wonbeeny/milkt-hme-v1"
token = {your_token}

configs = [
    "MilkT-RQ1-Solved_History",
    "HME-Cold_Start",
    "MilkT-Concept_List",
    "MilkT-Knowledge_Map",
    "MilkT-Question_Text",
    "Total-Question_Meta"
]

for file_name in configs:
    local_dir = snapshot_download(
        repo_id=repo_id,
        repo_type="dataset",
        allow_patterns=[f"{file_name}/*.parquet"],   # 필요한 config만
        token=token,
    )
    
    # 해당 폴더 안 parquet들 전부 읽기
    parquet_paths = Path(local_dir).glob(f"{file_name}/*.parquet")
    df = pd.concat([pd.read_parquet(path) for path in parquet_paths], ignore_index=True)
    if file_name == "HME-Cold_Start":
        df.to_csv(f"./Offline/{file_name}.csv", index=False)
    else:
        df.to_csv(f"./Online/{file_name}.csv", index=False)
```
